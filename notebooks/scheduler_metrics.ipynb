```python
import json
from pathlib import Path
def load_metrics(name):
    data = json.loads(Path(f'/tmp/{name}_summary.json').read_text())
    npu = data.get('npu_metrics', {})
    return {
        'config': name,
        'mips': data.get('mips'),
        'utilization': npu.get('utilization'),
        'avg_wait_cycles': npu.get('avg_wait_cycles'),
    }
rows = [load_metrics(name) for name in ['rr_baseline', 'priority_tuning']]
rows
```
